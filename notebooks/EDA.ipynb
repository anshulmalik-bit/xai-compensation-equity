Validates the synthetic data distribution and shows the unadjusted gender pay gap.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

df = pd.read_csv('../data/india_compensation_dataset.csv')
sns.set_theme(style="whitegrid")
```

## Salary Distribution

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['ctc'], bins=50, kde=True)
plt.title('CTC Distribution (LPA)')
plt.xlabel('CTC (LPA)')
plt.ylabel('Count')
plt.show()
```

## Unadjusted Gender Pay Gap by Level

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x='job_level', y='ctc', hue='gender')
plt.title('Unadjusted Salary by Job Level & Gender')
plt.xlabel('Job Level')
plt.ylabel('CTC (LPA)')
plt.show()
```
